# Data Validation: Multimodal Traffic Counts
* **Dataset:** comptage_multimodal_raw.json
* **Checks:** File existence, non-zero file size, and valid non-empty JSON structure.

```python
import json
import os

path = "/Volumes/workspace/default/raw_data/comptage_multimodal_raw.json"

# 1. Check file exists and is not empty
assert (
    os.path.exists(path) and os.path.getsize(path) > 100
), "File missing or empty!"  
```

```python
# 2. Check JSON is valid and contains records
with open(path, "r") as f:
    data = json.load(f)
    assert len(data) > 0, "JSON payload is empty!"

print(f"Validation successful: {len(data)} records ready for Bronze.")
```
