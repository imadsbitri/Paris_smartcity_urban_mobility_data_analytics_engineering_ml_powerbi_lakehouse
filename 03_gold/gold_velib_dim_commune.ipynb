# Commune Dimension Data - Gold Layer

## Objective
Extract and deduplicate unique commune attribute data from silver.silver_velib to build a standardized Commune Dimension Gold Delta table (gold.gold_dim_commune).

## Data Flow
silver.silver_velib → Spark SQL / DataFrame → gold.gold_dim_commune

## Source
The underlying data comes from the Paris OpenData API: Vélib' - Emplacements des stations.

## Input
Silver Delta table: silver.silver_velib

## Output
Gold Delta table: gold.gold_dim_commune

## Gold Layer Principle
The Gold layer delivers curated, dimensional models and business-level aggregations ready for reporting and analytics. This pipeline isolates unique geographical communes and INSEE codes to maintain a clean dimension table with strict entity integrity.

## Processing Steps
1. **Load Silver Data:** Read silver.silver_velib into PySpark and display initial dataset.
2. **Extract Dimension:** Query distinct commune names and INSEE codes.
3. **Write to Gold:** Persist deduplicated dataset to gold.gold_dim_commune Delta table.

```python
# Load data from silver schema
df_silver_velib=spark.table('workspace.silver.silver_velib')
```

```python
# display the dataframe 
df_silver_velib.display()
```

## BUSINESS TRANSFORMATION AND MODELING

```python
# Extract Commune Dimension Data
query_dim_commune = """
SELECT DISTINCT 
    nom_arrondissement_communes AS commune_name, 
    code_insee_commune AS code_insee
FROM silver.silver_velib
"""

df_dim_commune = spark.sql(query_dim_commune)
```

```python
# Display df_dim_commune
df_dim_commune.display()
```

# WRITING GOLD TABLE

```python
# writing df_dim_commune to gold schema
df_dim_commune\
    .write\
        .format("delta")\
        .mode("overwrite")\
        .option("overwriteSchema", "true") \
        .saveAsTable("gold.gold_velib_dim_commune")

```
